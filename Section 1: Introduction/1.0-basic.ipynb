### Indentation
Indentation is used to define the structure & hierarchy of the code. Unlike many other programming languages that use braces {} to delimit blocks of code, python uses indentation to determine the grouping of statements. This means that all the statements within a block must be indented at the same level.

```python
## Indentation
## Python uses indentation to define blocks of code. Consistent use of spaces (commonly 4) or a tab is required.

age=32
if age>30:
    print(age)

print(age)
```
