### DataTypes
### 1.Definition
=> Data types are a classfication of data which tell the compiler or interpreter how the programmer intends to use the data.

=> They determine the type of operations that can be performed on the data, the values that the data can take, and the amoount of memory needed to store the data.

### 2. Importance of data types in programing:

### Explanation:
=> Data types ensure that data is stored in an efficient way.

=> They help in performing correct iperations on data.

=> Proper use of data types can prevvent errors & bugs in the program.
