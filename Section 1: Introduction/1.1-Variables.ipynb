### Variables
Variables are fundamental elements in programing used to storeed data that can be referenced and manipulated in a program. In python, variables are created when you assign a value to them, and they do not need explilcit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable.

### Explicit Declaration:
An explicit decalration means clearly stating or defining something directly, leaving no room for guesswork or automatic inference.

### In Programming:
An explicit declaration is when you clearly specify a variable, tupe, or behavior instead of letting the system infer it.

### Example In Python:
x=5 #Implicit: Python infers x is an integer.
x: int=5 #Explicit: You clearly declear x is an integer.
### In General English:
"Explicit" means clearly & directly expressed, without ambiguity.
### Explicit Declaration: 
You say exactly what something is.
### Opposit -> Implicit Declaration:
The meaning is assumed or inferred.
### Why Use Expliit Declaration:
1. Improves readability.
2. Helps avoid bugs.
3. useful in large codebases or collaboration.
