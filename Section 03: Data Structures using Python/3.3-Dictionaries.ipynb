### Introduction to Dictionary

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g., srings, numbers, or tuples), while values can be any type.

```python
## Creating Dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"Waqar", "age":28, "grade": 75}
print(student)
print(type(student))
```

```python
## Single key is always used
student={"name":"Waqar", "age":28, "name": 75}
print(student)
```

```python
### Accessing Dictionary Elements
student={"name":"Waqar", "age":28, "grade": "A"}
print(student)
```

```python
### Accessing Dictionary Elements
print(student['grade'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available")) # Not Available is by default value
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable, so you can add, update or delete elements
print(student)
```

```python
student["age"]=33 ## update value for the key
print(student)
student["address"]="Pakistan" ## added a new key and value
print(student)
```

```python
del student['Address'] ## delete key and value pair
print(student)
```

```python
## Dictionary methods

keys=student.keys() ## get all the keys
print(keys)
values=student.values() # get all values
print(values)

items=student.items() # get all key value pairs
print(items)
```

```python
## Shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="Waqar1"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## shallow copy
print(student_copy1)
print(student)
```

```python
student["name"]="Waqar2"
print(student_copy1)
print(student)
```

# What is a Shallow Copy in Python?

A shallow copy is a copy of an object where the outer object (like a list or dictionary) is duplicated, but the nested or inner objects are still shared between the original and the copy.

Think of it like copying a list of boxes — you get a new list, but the same boxes inside it.

A shallow copy creates a new outer container, but reuses references to the inner objects.

```python
## Iterating over dictionaries
## You can use loops to iterate over dictionaries, Keys, values or items

## iterating over keys
for key in student.keys():
    print(key)
```

```python
### Iterate over values
for value in student.values():
    print(value)
```

```python
### Iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Dictionaries
students={
    "student1": {"name":"Waqar", "age": 23},
    "student2": {"name":"Waqas", "age": 27},
}
print(students)
```

```python
## Access nested dictionaries elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Condition dictionary comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practicle Examples

## use a dictionary to count the frequency of elements in list
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 dictionaries into one

dict1={"a":1, "b":2}
dict2={"b":3, "c":4}
merged_dict={**dict1, **dict2} # => Keywords argument
print(merged_dict)
```
