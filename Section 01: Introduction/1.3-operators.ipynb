```python
### What is the difference between floor division and modulus operation

###  **Floor Division (`//`)** vs **Modulus (`%`)**

#These are both arithmetic operations used with numbers

### 1. **Floor Division (`//`)**

# Returns the quotient of a division, but rounded down to the nearest whole number.
# It discards the remainder.

#### Example:
7 // 3   # → 2  (because 3 goes into 7 two times)


### 2. **Modulus (`%`)**

# Returns the remainder of a division.

#### Example:
7 % 3   # → 1  (because 3 × 2 = 6, and 7 - 6 = 1)

### Combined Example:
a = 7
b = 3

quotient = a // b  # 2
remainder = a % b  # 1

# So:

# 7 divided by 3 is:
#→ 2 quotient (floor division)
#→ 1 remainder (modulus)

# Because:

# 7 = 3 * 2 + 1

### Use Cases:

#| Operation           | Use Case Example                             |
#| // Floor Division | Splitting things evenly (e.g. pages, seats)     |
#| % Modulus         | Finding remainders (e.g. even/odd check: x % 2) |

#Basic Definition:
#Quotient = How many times the divisor fits into the dividend.
#Dividend ÷ Divisor = Quotient + (Remainder)
#12 ÷ 4 = 3

#12 is the dividend
#4 is the divisor
#3 is the quotient
```
