### Introduction to Funtions

Definition:

A funtion is a block of code that performs a specific task. Functions help in organizing code, reusing code, and improving readability.

```python
### Syntax
def function_name(parameters):
    """Docstring"""
    # Function body
    return expression
```

```python
## Why functions?
num=24
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
def even_or_odd(num):
    """This function prerform finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is even")
```

```python
# Call this function
even_or_odd(24)
```

```python
## function with multiple parameters

def add(a,b):
    return a+b

result = add(2,4)
print(result)
```

```python
## Default Parameters

def greet(name):
    print(f"Hello {name} Welcome to the paradise")

greet("Waqar")
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome to the paradise")

greet()
```

```python
### Variable Length Arguments
## Positional and keyword arguments

def print_numbers(*waqar): # Here * referenced by all positional arguments
    for number in waqar:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Waqar")
```

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Waqar")
```

```python
## Keywords arguments

def print_details(**kwargs): # **kwargs means key value pairs
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Waqar", age="35", country="Pakistan") ## All these values referenced by **kwargs
```

```python
def print_details(*args, **kwargs): # **kwargs means key value pairs
    for val in args:
        print(f"Positional argument {val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Waqar",name="Waqar", age="35", country="Pakistan")
```

```python
### Return Statements

def multiply(a,b):
    return a*b

multiply(2,3)
```

```python
### Return multiple parameters

def multiply(a,b):
    return a*b,a

multiply(2,3)
```
