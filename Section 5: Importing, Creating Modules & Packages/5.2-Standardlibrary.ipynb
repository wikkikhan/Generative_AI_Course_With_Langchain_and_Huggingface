### Standard Library Overview
Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range fo functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.

```python
import array
arr=array.array('i',[1,2,3,4])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random

import random
print(random.randint(1,10))
print(random.choice(['apple', 'banana', 'cherry']))
```

```python
### File And Directory Access

import os
print(os.getcwd())
```

```python
os.mkdir('test_dir')
```

```python
## High Level operations on files and collection of files
import shutil
shutil.copyfile('source.txt','destination.text')
```

```python
## Data Serialization
import json
data={'name':"Waqar",'age':25}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## CSV

import csv

with open('example.csv', mode='w', newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Waqar', 28])

with open('example.csv', mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime, timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```

```python
## time
import time

print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular expression
import re

pattern=r'\d+'
text='There are 123 apples 456'
match=re.search(pattern, text)
print(match.group())
```
