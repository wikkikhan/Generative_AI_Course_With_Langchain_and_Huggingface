```python
### List comprehension
## List comprehension is a short and elegant way to create lists in Python.
   ## => It lets you write less code to build a new list by looping through something

### Exmple of Basic for Loop
lst=[]
for x in range(10):
    lst.append(x**2)

print(lst)
```

```python
### Using List Comprehension

## Basic Syntax:
## [expression for item in iterable if condition]
    # expression → What to put in the list
    # item → Variable name for each element
    # iterable → Something you can loop over (like a list or range)
    # condition (optional) → Filters which items to include


[x**2 for x in range(10)]
```

```python
### Conclusion:
#| Benefit          | Explanation                      |
#| ---------------- | -------------------------------- |
#| Shorter Code     | Less lines, more readable        |
#| Faster Execution | Often faster than loops          |
#| Clear Intent     | Easy to see what you're building |
```

## List Comprehension

Basic Syntax => [expression for itme in iterable]

with conditional logic [expression for itme in iterable if condition]

Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
## Basic List Comprehension

square=[num**2 for num in range(10)]
print(square)
```

```python
## List Comprehension with condition

lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)
print(lst)
```

```python
even_numbers=[num for num in range(10) if num%2==0]
print(even_numbers)
```

```python
#3 Nested List Comprehension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j] for i in lst1 for j in lst2]

print(pair)
```

```python
## List Comprehension with function calls
words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths) # Output [5, 5, 6, 4, 13]
```

### Conclusion
