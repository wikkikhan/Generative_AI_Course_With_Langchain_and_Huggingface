### Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after limmatization is called 'lemma', which is a root word rather than a root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader to find a lemma. Let us understand it with an example.

```python
### Some of the use cases Q&A, chatbots, text summarization
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS - Noun-n
verb-v
adjective-a
adverb-r
'''

lemmatizer.lemmatize('going', pos='v')
```

```python
words = ["eating", "eats", "eaten", "writing", "writes", "programming", "program", "history", "finally", "finalize"]
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word, pos='v'))
```

```python
lemmatizer.lemmatize('goes', pos='v')
```

```python
lemmatizer.lemmatize('fairly', pos='v'),lemmatizer.lemmatize('sportingly', pos='v')
```
