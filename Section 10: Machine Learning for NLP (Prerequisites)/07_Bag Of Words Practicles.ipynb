```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
import string    
```

```python
## Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
```

```python
## Sample Paragraph
paragraph="""
Natural Language Processing is a field of Artificial Intelligence. 
It helps machines understand human language. It is used in chatbots, 
language translation, and sentiment analysis.
"""
```

```python
# step1: Sentence tokenization
sentences = sent_tokenize(paragraph)
```

```python
sentences
```

```python
## step2: Word Tokenization + Preprocessing
## Initialize empty vocabulary and list of word lists
vocab = set()
processed_sentences = []

for sentence in sentences:
    words = word_tokenize(sentence.lower()) ## lower + tokenize
    words = [word for word in words if word not in string.punctuation] ## Remove Punctuation
    words = [word for word in words if word not in stopwords.words('english')] ## Remove Stopwords
    vocab.update(words)
    processed_sentences.append(words) ## Save Cleaned Senetence
```

```python
vocab
```

```python
processed_sentences
```

```python
## Step3: Create BoW Matrix
vocab = sorted(list(vocab)) ## Sort Vocabulary
bow_matrix = []

for sentence in processed_sentences:
    vector = []
    for word in vocab:
        vector.append(sentence.count(word))
    bow_matrix.append(vector)
```

```python
## Step4: Display BoW Output
print("Vocabulary: ", vocab)
print("\n Bag of Words Matrix:")
for row in bow_matrix:
    print(row)
```
